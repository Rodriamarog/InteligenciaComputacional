```python
from openai import OpenAI
import os
from google.colab import userdata

SYSTEM_MESSAGE = "You are a chatbot. You will have a conversation with a user. Be friendly and concise"

if __name__ == "__main__":
    KEY = userdata.get('OPENAI_API_KEY')
    MODEL = userdata.get('OPENAI_TEXT_MODEL')

    # Basic validation for secrets
    if not KEY or not MODEL:
        print("Error: OPENAI_API_KEY or OPENAI_TEXT_MODEL not found in Colab Secrets. Please add them.")
    else:
        client = OpenAI(
            api_key=KEY,
        )

        print(f"Chatting with {MODEL} model\n")


        conversation_history = [{'role': 'system', 'content': SYSTEM_MESSAGE}]

        while True:
            message = input("> ")
            conversation_history.append({'role': 'user', 'content': message})

            response = client.chat.completions.create(
                model=MODEL,
                messages=conversation_history
            )
            assistant_response = response.choices[0].message.content
            print(assistant_response)

            conversation_history.append({'role': 'assistant', 'content': assistant_response})
```
